```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
import numpy as np

# 1. Create dummy image data (50 sample images of size 224x224x3)
# Replace 'your_images' and 'your_labels' with your real dataset later
your_images = np.random.randint(0, 256, size=(50, 224, 224, 3), dtype=np.uint8)
your_labels = np.random.randint(0, 2, size=(50,))  # 0 for healthy, 1 for diseased

# Preprocess image pixel values to match MobileNetV2 expectations (-1 to 1)
your_images = preprocess_input(your_images.astype(np.float32))

# 2. Load pre-trained MobileNetV2 base model without the top layer
base = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# 3. Freeze base layer weights (transfer learning)
base.trainable = False

# 4. Build the custom classification head
model = keras.Sequential([
    base,
    layers.GlobalAveragePooling2D(),
    layers.Dense(2, activation="softmax")  # 2 output classes: healthy vs. diseased
])

# 5. Compile the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 6. Train the top layers
model.fit(your_images, your_labels, epochs=10, batch_size=8)
```
