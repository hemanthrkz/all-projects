```python
# ===============================
# NLP Preprocessing in Python
# ===============================

import re
from sklearn.feature_extraction.text import (
    CountVectorizer,
    TfidfVectorizer,
    ENGLISH_STOP_WORDS
)

# ==========================================
# 1. Text Preprocessing
# ==========================================

text = "Natural Language Processing is AMAZING, in 2026!"

# Convert to lowercase and remove numbers/special characters
clean = re.sub(r"[^a-z\s]", "", text.lower())

# Tokenization
tokens = clean.split()

# Remove stop words
tokens = [t for t in tokens if t not in ENGLISH_STOP_WORDS]

print("Original Text:")
print(text)

print("\nClean Text:")
print(clean)

print("\nTokens after removing stop words:")
print(tokens)

# ==========================================
# 2. Bag of Words (BoW)
# ==========================================

sentences = [
    "i love nlp",
    "nlp is fun",
    "i love ai"
]

cv = CountVectorizer()
bow = cv.fit_transform(sentences)

print("\nVocabulary:")
print(cv.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow.toarray())

# ==========================================
# 3. TF-IDF
# ==========================================

docs = [
    "the cat sat",
    "the dog ran",
    "the bird flew"
]

tv = TfidfVectorizer()
matrix = tv.fit_transform(docs)

words = dict(zip(tv.get_feature_names_out(), matrix.toarray()[0].round(2)))

print("\nTF-IDF Values (First Document):")
print(words)

print("\nWeight of 'the':", words["the"])
print("Weight of 'cat':", words["cat"])
```
