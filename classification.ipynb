```python
# Import libraries
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Dataset
data = [
    ("Congratulations! You won ₹10,000. Claim now!", "spam"),
    ("Call me when you reach home.", "ham"),
    ("Limited offer! Buy now!", "spam"),
    ("Let's meet tomorrow at 10 AM.", "ham"),
    ("Your OTP is 123456.", "ham"),
    ("Free recharge available. Click here!", "spam"),

    # Add your own messages
    ("Happy Birthday! Have a wonderful day.", "ham"),
    ("Win an iPhone for free. Register today!", "spam"),
    ("Can we have lunch together?", "ham"),
    ("Exclusive offer! Get 90% discount now.", "spam")
]

texts = [x[0] for x in data]
labels = [x[1] for x in data]

cv = CountVectorizer()
X = cv.fit_transform(texts)

model = MultinomialNB()
model.fit(X, labels)

# Test messages
test = [
    "Free recharge today",
    "See you tomorrow",
    "Win cash prize",
    "Good Morning"
]

X_test = cv.transform(test)

prediction = model.predict(X_test)

for msg, pred in zip(test, prediction):
    print(msg, "->", pred)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

data = [
    ("Congratulations! You won ₹10,000", "spam"),
    ("Let's meet tomorrow", "ham"),
    ("Free recharge available", "spam"),
    ("How are you?", "ham"),
    ("Claim your reward now", "spam"),
    ("Happy Birthday", "ham")
]

texts = [x[0] for x in data]
labels = [x[1] for x in data]

cv = CountVectorizer()
X = cv.fit_transform(texts)

model = LogisticRegression(max_iter=1000)

model.fit(X, labels)

test = [
    "Free money",
    "Good evening",
    "Claim prize now",
    "Call me"
]

X_test = cv.transform(test)

prediction = model.predict(X_test)

for msg, pred in zip(test, prediction):
    print(msg, "->", pred)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

reviews = [
    ("This movie is amazing", "positive"),
    ("I love this phone", "positive"),
    ("Excellent service", "positive"),
    ("Very bad experience", "negative"),
    ("Worst product", "negative"),
    ("I hate this", "negative"),
    ("Not good", "negative"),
    ("Not bad", "positive")
]

texts = [x[0] for x in reviews]
labels = [x[1] for x in reviews]

tfidf = TfidfVectorizer(ngram_range=(1,2))

X = tfidf.fit_transform(texts)

model = LogisticRegression(max_iter=1000)
model.fit(X, labels)

test_reviews = [
    "This phone is amazing",
    "Worst movie",
    "Not good",
    "Not bad",
    "Excellent product",
    "Very bad"
]

X_test = tfidf.transform(test_reviews)

prediction = model.predict(X_test)

for review, pred in zip(test_reviews, prediction):
    print(review, "->", pred)
```
